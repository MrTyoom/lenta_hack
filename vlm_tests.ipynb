```python
from openai import OpenAI

import base64
import glob
import os

PATH_TO_CROPS = 'data/'
```

```python
!vllm serve OpenGVLab/InternVL3_5-8B \
  --trust-remote-code \
  --dtype float16 \
  --max-model-len 1024 \
  --gpu-memory-utilization 0.72 \
  --limit-mm-per-prompt image=1 \
  --enforce-eager
```

```python
client = OpenAI(
    api_key="EMPTY",
    base_url="http://localhost:8000/v1",
)

with open(os.path.join(PATH_TO_CROPS, 'crop1.png'), "rb") as f:
    image_b64 = base64.b64encode(f.read()).decode()

response = client.chat.completions.create(
    model="AvitoTech/avision",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "You are OCR system. Read Russian store price tag. Return ONLY in this format: PRODUCT: ..., PRICE: ..."},
                {
                    "type": "image_url",
                    "image_url": {
                        "url": f"data:image/png;base64,{image_b64}"
                    },
                },
            ],
        }
    ],
    max_tokens=90,
)

print(response.choices[0].message.content)
```
